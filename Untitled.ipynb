# Multifield fluctuation visualization

`runs/` からrunを選び、場のゆらぎ、自己・クロススペクトル、曲率スペクトル、
ヒストグラム、最新2Dスライスを可視化します。

1. `RUN_INDEX` を選ぶ（`-1` は最新run）
2. 必要なら `SAVE_FIGURES = True` にする
3. **Kernel → Restart Kernel and Run All Cells** を実行する

大きなスペクトル／ヒストグラムCSVはチャンク単位で読み、最新の `output_id` だけを
メモリへ残します。CSV中の `NaN` は未定義値なので0には置換しません。


```python
%matplotlib inline

from pathlib import Path
import math
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({"figure.dpi": 120, "savefig.dpi": 200})
pd.set_option("display.max_columns", 50)

# Trueにすると、各図を選択runの figures/ にPNG保存します。
SAVE_FIGURES = False
```

## 1. runを選択


```python
def find_project_root(start: Path) -> Path:
    start = start.resolve()
    for candidate in (start, *start.parents):
        if (candidate / "runs").is_dir() and (candidate / "Project.toml").is_file():
            return candidate
    raise FileNotFoundError(
        "Project.toml と runs/ があるMultifieldInflationLatticeディレクトリから "
        "Notebookを開いてください。"
    )


PROJECT_ROOT = find_project_root(Path.cwd())
RUNS_ROOT = PROJECT_ROOT / "runs"
run_dirs = sorted(
    [path for path in RUNS_ROOT.iterdir()
     if path.is_dir() and (path / "background.csv").is_file()],
    key=lambda path: (path / "background.csv").stat().st_mtime,
)
if not run_dirs:
    raise FileNotFoundError(f"background.csvを含むrunがありません: {RUNS_ROOT}")

run_table = pd.DataFrame([
    {
        "index": index,
        "run_id": path.name,
        "background_MiB": (path / "background.csv").stat().st_size / 2**20,
        "updated": pd.Timestamp.fromtimestamp((path / "background.csv").stat().st_mtime),
    }
    for index, path in enumerate(run_dirs)
])
display(run_table)
```

```python
# 上の表のindexを指定します。-1なら最新runです。
RUN_INDEX = -1
RUN = run_dirs[RUN_INDEX]
print(f"Selected run: {RUN}")
```

## 2. データ読込み


```python
def read_required_csv(filename: str) -> pd.DataFrame:
    path = RUN / filename
    if not path.is_file():
        raise FileNotFoundError(f"必要な出力がありません: {path}")
    return pd.read_csv(path)


def read_latest_output(path: Path, usecols=None, chunksize: int = 250_000) -> pd.DataFrame:
    # 大きなCSVから最大output_idの行だけをチャンク読込みする。
    if not path.is_file() or path.stat().st_size == 0:
        return pd.DataFrame()
    columns = None if usecols is None else list(dict.fromkeys(["output_id", *usecols]))
    latest_id = None
    selected = []
    try:
        reader = pd.read_csv(path, usecols=columns, chunksize=chunksize)
        for chunk in reader:
            if chunk.empty:
                continue
            ids = pd.to_numeric(chunk["output_id"], errors="coerce")
            if not ids.notna().any():
                continue
            chunk_latest = int(ids.max())
            rows = chunk.loc[ids == chunk_latest].copy()
            if latest_id is None or chunk_latest > latest_id:
                latest_id = chunk_latest
                selected = [rows]
            elif chunk_latest == latest_id:
                selected.append(rows)
    except pd.errors.EmptyDataError:
        return pd.DataFrame()
    return pd.concat(selected, ignore_index=True) if selected else pd.DataFrame()


def present(fig, filename: str):
    fig.tight_layout()
    if SAVE_FIGURES:
        directory = RUN / "figures"
        directory.mkdir(exist_ok=True)
        destination = directory / filename
        fig.savefig(destination, bbox_inches="tight")
        print(f"Saved: {destination}")
    plt.show()


background = read_required_csv("background.csv")
energies = read_required_csv("energies.csv")
slowroll = read_required_csv("slowroll.csv")
diagnostics = read_required_csv("diagnostics.csv")
field_names = [column[:-len("_variance")]
               for column in background.columns if column.endswith("_variance")]

print(f"Rows: background={len(background):,}, energies={len(energies):,}")
print(f"Fields: {field_names}")
display(background.tail(3))
```

## 3. ゆらぎ振幅の時間発展


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

for field in field_names:
    variance = pd.to_numeric(background[f"{field}_variance"], errors="coerce").clip(lower=0)
    sigma = np.sqrt(variance)
    mean_abs = pd.to_numeric(background[f"{field}_mean"], errors="coerce").abs()
    relative = sigma / mean_abs.replace(0, np.nan)

    axes[0].plot(background["efolds"], sigma, label=fr"$\sigma_{{{field}}}$")
    axes[1].plot(background["efolds"], relative, label=fr"$\sigma_{{{field}}}/|\bar{{{field}}}|$")

axes[0].set(
    xlabel="e-folds N",
    ylabel="spatial standard deviation",
    title="Absolute field fluctuations",
    yscale="log",
)
axes[1].set(
    xlabel="e-folds N",
    ylabel="relative standard deviation",
    title="Relative field fluctuations",
    yscale="log",
)
for axis in axes:
    axis.legend()

present(fig, "field-fluctuation-amplitudes.png")
```

## 4. 最新時刻の場パワースペクトル


```python
spectrum_columns = [
    "step", "efolds", "k_comoving", "k_physical", "n_modes",
    "field_i", "field_j", "power_dimensionless",
]
field_spectra = read_latest_output(
    RUN / "field_spectra.csv", usecols=spectrum_columns,
)
if field_spectra.empty:
    raise ValueError("field_spectra.csvに可視化可能な行がありません。")

field_spectra = field_spectra.replace([np.inf, -np.inf], np.nan)
field_spectra = field_spectra[field_spectra["k_comoving"] > 0]
auto = field_spectra[field_spectra["field_i"] == field_spectra["field_j"]]
cross = field_spectra[field_spectra["field_i"] != field_spectra["field_j"]]

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
for (field_i, field_j), rows in auto.groupby(["field_i", "field_j"], dropna=False):
    rows = rows.sort_values("k_comoving")
    valid = rows["power_dimensionless"] > 0
    axes[0].loglog(
        rows.loc[valid, "k_comoving"], rows.loc[valid, "power_dimensionless"],
        marker="o", markersize=3, label=f"{field_i} auto",
    )

for (field_i, field_j), rows in cross.groupby(["field_i", "field_j"], dropna=False):
    rows = rows.sort_values("k_comoving")
    axes[1].plot(
        rows["k_comoving"], rows["power_dimensionless"],
        marker="o", markersize=3, label=f"{field_i} × {field_j}",
    )

axes[0].set(
    xlabel=r"comoving $k$",
    ylabel=r"dimensionless power $\Delta^2$",
    title="Field auto spectra",
)
axes[1].set(
    xlabel=r"comoving $k$",
    ylabel=r"dimensionless cross power $\Delta^2_{ij}$",
    title="Field cross spectra",
    xscale="log",
    yscale="symlog",
)
for axis in axes:
    if axis.lines:
        axis.legend()

output_id = int(field_spectra["output_id"].iloc[0])
efolds = float(field_spectra["efolds"].iloc[0])
fig.suptitle(f"output_id={output_id}, N={efolds:.6g}")
present(fig, "latest-field-spectra.png")
```

## 5. 線形曲率と非線形δNスペクトル


```python
curvature_columns = [
    "step", "efolds", "k_comoving", "k_physical", "n_modes",
    "component", "field_i", "field_j", "power_dimensionless",
]
linear_spectra = read_latest_output(
    RUN / "linear_curvature_spectra.csv", usecols=curvature_columns,
)
linear_total = linear_spectra[linear_spectra.get("component") == "total"].copy()
linear_total = linear_total.replace([np.inf, -np.inf], np.nan).dropna(
    subset=["k_comoving", "power_dimensionless"]
)
linear_total = linear_total[
    (linear_total["k_comoving"] > 0) & (linear_total["power_dimensionless"] > 0)
].sort_values("k_comoving")

delta_path = RUN / "deltaN_spectrum.csv"
try:
    delta_spectrum = pd.read_csv(delta_path) if delta_path.is_file() else pd.DataFrame()
except pd.errors.EmptyDataError:
    delta_spectrum = pd.DataFrame()

fig, axis = plt.subplots(figsize=(7.5, 5))
if not linear_total.empty:
    axis.loglog(
        linear_total["k_comoving"], linear_total["power_dimensionless"],
        marker="o", markersize=3, label=r"linear $\zeta$",
    )
if not delta_spectrum.empty:
    valid = (
        np.isfinite(delta_spectrum["k_comoving"])
        & np.isfinite(delta_spectrum["power_dimensionless"])
        & (delta_spectrum["k_comoving"] > 0)
        & (delta_spectrum["power_dimensionless"] > 0)
    )
    axis.loglog(
        delta_spectrum.loc[valid, "k_comoving"],
        delta_spectrum.loc[valid, "power_dimensionless"],
        marker="s", markersize=3, label=r"nonlinear $\delta N$",
    )

axis.set(
    xlabel=r"comoving $k$",
    ylabel=r"dimensionless curvature power $\Delta_\zeta^2$",
    title="Curvature spectra",
)
if axis.lines:
    axis.legend()
present(fig, "curvature-spectra.png")

summary_path = RUN / "deltaN_summary.csv"
if summary_path.is_file() and summary_path.stat().st_size > 0:
    try:
        delta_summary = pd.read_csv(summary_path)
        if not delta_summary.empty:
            display(delta_summary.T.rename(columns={0: "deltaN summary"}))
    except pd.errors.EmptyDataError:
        pass
```

## 6. 最新ヒストグラム


```python
histogram_dir = RUN / "histograms"
histogram_paths = sorted(histogram_dir.glob("*.csv")) if histogram_dir.is_dir() else []

if not histogram_paths:
    print("ヒストグラム出力がありません。")
else:
    columns = ["quantity", "bin_left", "bin_right", "bin_center", "count", "density"]
    histogram_data = []
    for path in histogram_paths:
        latest = read_latest_output(path, usecols=columns)
        if not latest.empty:
            histogram_data.append((path.stem, latest))

    ncols = 2
    nrows = math.ceil(len(histogram_data) / ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(12, 3.8 * nrows), squeeze=False)
    for axis, (name, rows) in zip(axes.flat, histogram_data):
        rows = rows.sort_values("bin_center")
        total_count = pd.to_numeric(rows["count"], errors="coerce").sum()
        weighted_mean = (
            np.average(rows["bin_center"], weights=rows["count"])
            if np.isfinite(total_count) and total_count > 0 else 0.0
        )
        centered = rows["bin_center"] - weighted_mean
        valid = (
            np.isfinite(centered)
            & np.isfinite(rows["density"])
            & (rows["density"] > 0)
        )
        axis.semilogy(
            centered.loc[valid], rows.loc[valid, "density"],
            drawstyle="steps-mid",
        )
        axis.ticklabel_format(axis="x", style="sci", scilimits=(0, 0), useOffset=False)
        axis.set(
            xlabel=f"{name} − weighted mean",
            ylabel="probability density",
            title=f"δ{name}",
        )
    for axis in axes.flat[len(histogram_data):]:
        axis.set_visible(False)
    present(fig, "latest-histograms.png")
```

## 7. 最新2Dスライスのゆらぎマップ


```python
slice_dir = RUN / "slices"
pattern = re.compile(r"^(?P<quantity>.+)_(?P<output_id>\d{8})_(?P<plane>[xyz]\d+)$")
latest_slices = {}

if slice_dir.is_dir():
    for path in slice_dir.glob("*.csv"):
        match = pattern.match(path.stem)
        if match is None:
            continue
        quantity = match.group("quantity")
        output_id = int(match.group("output_id"))
        previous = latest_slices.get(quantity)
        if previous is None or output_id > previous[0]:
            latest_slices[quantity] = (output_id, path)

if not latest_slices:
    print("2Dスライス出力がありません。configの output.save_2d_slices を確認してください。")
else:
    ordered = sorted(latest_slices.items())
    ncols = 2
    nrows = math.ceil(len(ordered) / ncols)
    fig, axes = plt.subplots(nrows, ncols, figsize=(12, 5 * nrows), squeeze=False)

    for axis, (quantity, (output_id, path)) in zip(axes.flat, ordered):
        data = pd.read_csv(path)
        grid = data.pivot_table(index="y", columns="x", values="value", aggfunc="first")
        grid = grid.sort_index().sort_index(axis=1)
        fluctuations = grid.to_numpy(dtype=float).copy()
        fluctuations -= np.nanmean(fluctuations)
        scale = np.nanmax(np.abs(fluctuations))
        if not np.isfinite(scale) or scale == 0:
            scale = 1.0
        image = axis.imshow(
            fluctuations,
            origin="lower",
            extent=[grid.columns.min(), grid.columns.max(), grid.index.min(), grid.index.max()],
            aspect="equal",
            cmap="RdBu_r",
            vmin=-scale,
            vmax=scale,
        )
        axis.set(
            xlabel="physical coordinate 1",
            ylabel="physical coordinate 2",
            title=f"{quantity} − spatial mean (output_id={output_id})",
        )
        fig.colorbar(image, ax=axis, label="fluctuation")

    for axis in axes.flat[len(ordered):]:
        axis.set_visible(False)
    present(fig, "latest-fluctuation-slices.png")
```

## 読み方

- `sqrt(field_variance)` は各時刻の空間標準偏差です。
- 場の自己スペクトルは非負ですが、クロススペクトルは負になり得るため `symlog` 表示です。
- `linear ζ` が `NaN` の時刻は、平均場速度が設定しきい値未満で未定義です。
- 2Dマップは保存値から空間平均を引いたゆらぎを、0中心の共通色尺度で表示します。
- `SAVE_FIGURES = True` にして全セルを再実行すると、PNGが選択runの `figures/` に保存されます。
